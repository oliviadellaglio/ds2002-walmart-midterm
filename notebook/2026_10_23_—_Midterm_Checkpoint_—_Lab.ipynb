### 1 — Your cleaning pipeline as functions

**TODO:** paste the cleaning code you have written so far. It should be functions, not loose cells, and it should cover at minimum: duplicates, `unit_price` to numeric (it comes in as text, some with a `$`), the negative `quantity` values with your decision applied, `timestamp` parsed, and `store_id` cleaned up so it will join to the store file.

Print the row count before and after. We want to see how many rows your decisions removed.

```python
import pandas as pd

df = pd.read_csv(
    "/content/ds2002-walmart-midterm/data/walmart_transactions.csv"
)

def clean_transactions(df):
    clean_df = df.copy()

    print("Original rows:", len(clean_df))

    clean_df = clean_df.drop_duplicates()
    print("After removing duplicates:", len(clean_df))

    clean_df["unit_price"] = pd.to_numeric(
        clean_df["unit_price"].astype(str).str.replace(
            "$", "", regex=False
        ).str.replace(",", "", regex=False),
        errors="coerce"
    )

    clean_df["quantity"] = pd.to_numeric(
        clean_df["quantity"], errors="coerce"
    )

    clean_df = clean_df[clean_df["quantity"] > 0]
    print("After removing invalid quantities:", len(clean_df))

    clean_df["timestamp"] = pd.to_datetime(
        clean_df["timestamp"],
        format="mixed",
        errors="coerce"
    )

    clean_df = clean_df.dropna(subset=["timestamp"])
    print("After removing missing timestamps:", len(clean_df))

    clean_df["store_id"] = (
        clean_df["store_id"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    for col in ["category", "sku"]:
        if col in clean_df.columns:
            clean_df[col] = (
                clean_df[col]
                .astype(str)
                .str.strip()
                .str.upper()
            )

    print("Final cleaned rows:", len(clean_df))
    print("Total rows removed:", len(df) - len(clean_df))

    return clean_df


clean_df = clean_transactions(df)

clean_df.head()
```

### 4 — One working weather API call

**TODO:** show the call and the parsed DataFrame. It must include the retry logic and a `source` column. Say which timezone the weather timestamps are in and how you will line them up with the sales timestamps. If it is not working, paste the actual error and what you have tried.

```python
import pandas as pd
import numpy as np
import requests
import time

BASE = 'https://archive-api.open-meteo.com/v1/archive'
RETRY_STATUSES = {429, 500, 502, 503, 504}

def daily_weather(start, end, lat=28.54, lon=-81.38, tries=3):
    params = {
        'latitude': lat,
        'longitude': lon,
        'start_date': start,
        'end_date': end,
        'daily': 'precipitation_sum,windspeed_10m_max',
        'timezone': 'America/New_York',
    }

    for attempt in range(1, tries + 1):
        try:
            r = requests.get(BASE, params=params, timeout=10)

        except (requests.Timeout, requests.ConnectionError):
            if attempt == tries:
                raise
            time.sleep(2 ** attempt)
            continue

        if r.status_code in RETRY_STATUSES:
            if attempt == tries:
                r.raise_for_status()
            time.sleep(2 ** attempt)
            continue

        r.raise_for_status()

        frame = pd.DataFrame(r.json()['daily'])
        frame = frame.rename(columns={'time': 'date'})
        frame['date'] = pd.to_datetime(frame['date'])
        frame['source'] = 'open-meteo'

        return frame


try:
    weather_live = daily_weather(
        '2024-09-01',
        '2024-09-30'
    )

    print("Weather API call successful!")
    print("Source:", weather_live['source'].iloc[0])
    print("Number of days:", len(weather_live))

    display(weather_live.head())

except requests.RequestException as err:
    print("Weather API call failed:", err)
```

The weather data uses the America/New_York timezone. We plan to use the weather data and align it with Walmart transactions data. To do this, we have to convert the sales timestamps to calendar dates and join the datasets on date. We will also account for store locations when expanding the weather analysis.
